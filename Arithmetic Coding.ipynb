```python
# Arithmetic Coding

def arithmetic_coding(text):
   
    symbols = sorted(set(text))
    total = len(text)

    probability = {}
    for symbol in symbols:
        probability[symbol] = text.count(symbol) / total

    low_range = 0.0
    ranges = {}

    for symbol in symbols:
        high_range = low_range + probability[symbol]
        ranges[symbol] = (low_range, high_range)
        low_range = high_range

    low = 0.0
    high = 1.0

    for symbol in text:
        range_width = high - low
        symbol_low, symbol_high = ranges[symbol]

        high = low + range_width * symbol_high
        low = low + range_width * symbol_low

    encoded_value = (low + high) / 2

    print("\nSymbol Probabilities:")
    for symbol in symbols:
        print(symbol, ":", probability[symbol])

    print("\nSymbol Ranges:")
    for symbol in symbols:
        print(symbol, ":", ranges[symbol])

    print("\nOriginal Text:", text)
    print("Final Interval:", (low, high))
    print("Encoded Value:", encoded_value)

text = input("Enter a string: ")

arithmetic_coding(text)
```
