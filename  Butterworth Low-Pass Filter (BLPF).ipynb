```python
# Butterworth Low-Pass Filter (BLPF)
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("image.jpg")

if image is None:
    raise FileNotFoundError("Image not found!")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

hsv_image = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)

H, S, I = cv2.split(hsv_image)

dft = np.fft.fft2(I)
dft_shift = np.fft.fftshift(dft)

rows, cols = I.shape
crow, ccol = rows // 2, cols // 2

D0 = 50
n = 2

u = np.arange(rows) - crow
v = np.arange(cols) - ccol
V, U = np.meshgrid(v, u)

D = np.sqrt(U**2 + V**2)

mask = 1 / (1 + (D / D0)**(2 * n))

filtered_dft = dft_shift * mask

idft_shift = np.fft.ifftshift(filtered_dft)
idft = np.fft.ifft2(idft_shift)

filtered_I = np.abs(idft)
filtered_I = np.clip(filtered_I, 0, 255).astype(np.uint8)

filtered_hsv = cv2.merge((H, S, filtered_I))

filtered_image = cv2.cvtColor(
    filtered_hsv, cv2.COLOR_HSV2RGB
)

plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(np.log(1 + np.abs(dft_shift)), cmap="gray")
plt.title("Frequency Spectrum")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(filtered_image)
plt.title("Butterworth Low-Pass Filtered Image")
plt.axis("off")

plt.tight_layout()
plt.show()

cv2.imwrite(
    "BLPF_filtered_image.jpg",
    cv2.cvtColor(filtered_image, cv2.COLOR_RGB2BGR)
)

print("Butterworth Low-Pass Filtering Completed Successfully!")
```
