```python
# Gaussian Filter
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread("image.jpg")

if image is None:
    raise FileNotFoundError("Image not found!")

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

hsv_image = cv2.cvtColor(image, cv2.COLOR_RGB2HSV)

H, S, I = cv2.split(hsv_image)

H_filtered = cv2.GaussianBlur(H, (15, 15), 0)
S_filtered = cv2.GaussianBlur(S, (15, 15), 0)
I_filtered = cv2.GaussianBlur(I, (15, 15), 0)

filtered_hsv = cv2.merge(
    (H_filtered, S_filtered, I_filtered)
)

filtered_image = cv2.cvtColor(
    filtered_hsv, cv2.COLOR_HSV2RGB
)

plt.figure(figsize=(14, 6))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(filtered_image)
plt.title("Gaussian Filtered Image")
plt.axis("off")

plt.tight_layout()
plt.show()

cv2.imwrite(
    "gaussian_filtered_image.jpg",
    cv2.cvtColor(filtered_image, cv2.COLOR_RGB2BGR)
)

print("Gaussian Filtering Completed Successfully!")
```
